```python
import pandas as pd
import numpy as np

# Load the correct sheet
df = pd.read_excel(
    "/content/SkillZeno_Task1_Sales_Data_Analysis.xlsx",
    sheet_name="Raw Data"
)

# Clean column names
df.columns = df.columns.astype(str).str.strip()

# Convert numeric columns
for col in ["Quantity", "Unit Price", "Discount Rate"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Fill missing numeric values
for col in ["Quantity", "Unit Price", "Discount Rate"]:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(0)

# Calculate net sales
df["Net Sales"] = (
    df["Quantity"] * df["Unit Price"] * (1 - df["Discount Rate"])
)

# Calculate statistics
print("TOTAL SALES ANALYSIS")
print("--------------------")
print("Total Orders:", df["Order ID"].nunique())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Net Sales:", round(df["Net Sales"].sum(), 2))

order_sales = df.groupby("Order ID")["Net Sales"].sum()
print("Average Order Value:", round(order_sales.mean(), 2))
print("Highest Order Sales:", round(order_sales.max(), 2))

print("\nSALES BY REGION")
print(df.groupby("Region")["Net Sales"].sum().sort_values(ascending=False))

print("\nSALES BY CHANNEL")
print(df.groupby("Sales Channel")["Net Sales"].sum().sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Replace any true nulls or string 'nan' representation in Region with 'Unknown'
df["Region"] = df["Region"].fillna("Unknown").astype(str).str.strip()
df["Region"] = df["Region"].replace({"nan": "Unknown", "Nan": "Unknown"})
df["Region"] = df["Region"].str.title()

# Re-calculate regional sales grouping
region_sales = df.groupby("Region")["Net Sales"].sum().sort_values(ascending=False)

# Plot the corrected regional breakdown
region_sales.plot(kind="bar", title="Sales by Region (Corrected)",
                  xlabel="Region", ylabel="Net Sales")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Strip leading and trailing whitespace from the 'Product' column to merge duplicates (e.g. 'Headphones ' and 'Headphones')
df["Product"] = df["Product"].astype(str).str.strip()

product_sales = df.groupby("Product")["Net Sales"].sum().sort_values(ascending=False)

product_sales.plot(kind="bar", title="Sales by Product (Corrected)",
                   xlabel="Product", ylabel="Net Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

monthly_sales = (
    df.dropna(subset=["Order Date"])
    .assign(Month=lambda x: x["Order Date"].dt.to_period("M"))
    .groupby("Month")["Net Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind="line", marker="o",
                   title="Monthly Sales Trend",
                   xlabel="Month", ylabel="Net Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Install Plotly and widgets support
%pip -q install plotly ipywidgets

import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output

# Prepare dropdown filters
regions = ["All"] + sorted(df["Region"].dropna().unique().tolist())
channels = ["All"] + sorted(df["Sales Channel"].dropna().unique().tolist())

region_filter = widgets.Dropdown(
    options=regions, value="All", description="Region:"
)

channel_filter = widgets.Dropdown(
    options=channels, value="All", description="Channel:"
)

output = widgets.Output()

def show_dashboard(change=None):
    with output:
        clear_output(wait=True)

        filtered = df.copy()

        if region_filter.value != "All":
            filtered = filtered[filtered["Region"] == region_filter.value]

        if channel_filter.value != "All":
            filtered = filtered[filtered["Sales Channel"] == channel_filter.value]

        total_sales = filtered["Net Sales"].sum()
        total_orders = filtered["Order ID"].nunique()
        total_units = filtered["Quantity"].sum()
        avg_order = (
            filtered.groupby("Order ID")["Net Sales"].sum().mean()
            if total_orders else 0
        )

        print("📊 INTERACTIVE SALES DASHBOARD")
        print("=" * 45)
        print(f"💰 Total Net Sales: ₹{total_sales:,.2f}")
        print(f"𣜃 Total Orders: {total_orders}")
        print(f"📦 Total Units Sold: {total_units:,.0f}")
        print(f"📈 Average Order Value: ₹{avg_order:,.2f}")

        if filtered.empty:
            print("No data available for these filters.")
            return

        region_sales = (
            filtered.groupby("Region", as_index=False)["Net Sales"].sum()
        )
        fig1 = px.bar(
            region_sales, x="Region", y="Net Sales",
            title="Net Sales by Region",
            text_auto=".2s"
        )
        fig1.show()

        product_sales = (
            filtered.groupby("Product", as_index=False)["Net Sales"].sum()
            .sort_values("Net Sales", ascending=False)
        )
        fig2 = px.bar(
            product_sales, x="Product", y="Net Sales",
            title="Net Sales by Product",
            text_auto=".2s"
        )
        fig2.show()

        channel_sales = (
            filtered.groupby("Sales Channel", as_index=False)["Net Sales"].sum()
        )
        fig3 = px.pie(
            channel_sales, names="Sales Channel", values="Net Sales",
            title="Sales Share by Channel"
        )
        fig3.show()

        dated = filtered.dropna(subset=["Order Date"]).copy()
        if not dated.empty:
            dated["Month"] = dated["Order Date"].dt.to_period("M").astype(str)
            monthly = (
                dated.groupby("Month", as_index=False)["Net Sales"].sum()
            )
            fig4 = px.line(
                monthly, x="Month", y="Net Sales",
                markers=True, title="Monthly Sales Trend"
            )
            fig4.show()

        if not region_sales.empty:
            print("\nMajor insight:")
            best_region = region_sales.loc[
                region_sales["Net Sales"].idxmax(), "Region"
            ]
            print(f"Highest-sales region in this selection: {best_region}")

region_filter.observe(show_dashboard, names="value")
channel_filter.observe(show_dashboard, names="value")

display(widgets.HBox([region_filter, channel_filter]))
display(output)
show_dashboard()
```

```python
# Clean and prepare the data
for col in ["Quantity", "Unit Price", "Discount Rate"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

df["Net Sales"] = (
    df["Quantity"] * df["Unit Price"] * (1 - df["Discount Rate"])
)

# Calculate business KPIs
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()
total_sales = df["Net Sales"].sum()
average_order_value = df.groupby("Order ID")["Net Sales"].sum().mean()

print("BUSINESS PERFORMANCE SUMMARY")
print("=" * 40)
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
print("Total Net Sales: ₹", round(total_sales, 2))
print("Average Order Value: ₹", round(average_order_value, 2))

print("\nSALES BY REGION")
print(
    df.groupby("Region")["Net Sales"]
    .sum().sort_values(ascending=False).round(2)
)

print("\nSALES BY CHANNEL")
print(
    df.groupby("Sales Channel")["Net Sales"]
    .sum().sort_values(ascending=False).round(2)
)

print("\nTOP 5 PRODUCTS BY NET SALES")
print(
    df.groupby("Product")["Net Sales"]
    .sum().sort_values(ascending=False).head(5).round(2)
)
```

```python
# 1. Sales by Region
region_sales = df.groupby("Region")["Net Sales"].sum().sort_values(ascending=False)

region_sales.plot(kind="bar", title="Business Sales by Region",
                  xlabel="Region", ylabel="Net Sales",
                  figsize=(8, 5))
plt.tight_layout()
plt.show()

# 2. Top Products by Sales
product_sales = df.groupby("Product")["Net Sales"].sum().sort_values(ascending=False)

product_sales.plot(kind="bar", title="Product Sales Performance",
                   xlabel="Product", ylabel="Net Sales",
                   figsize=(9, 5))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Monthly Sales Trend
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)
monthly_sales = df.groupby("Month")["Net Sales"].sum()

monthly_sales.plot(kind="line", marker="o",
                   title="Monthly Business Sales Trend",
                   xlabel="Month", ylabel="Net Sales",
                   figsize=(10, 5))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
