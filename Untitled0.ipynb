```python
import pandas as pd
import numpy as np

# Load the correct sheet
df = pd.read_excel(
    "/content/SkillZeno_Task1_Sales_Data_Analysis.xlsx",
    sheet_name="Raw Data"
)

# Clean column names
df.columns = df.columns.astype(str).str.strip()

# Convert numeric columns
for col in ["Quantity", "Unit Price", "Discount Rate"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Fill missing numeric values
for col in ["Quantity", "Unit Price", "Discount Rate"]:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(0)

# Calculate net sales
df["Net Sales"] = (
    df["Quantity"] * df["Unit Price"] * (1 - df["Discount Rate"])
)

# Calculate statistics
print("TOTAL SALES ANALYSIS")
print("--------------------")
print("Total Orders:", df["Order ID"].nunique())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Net Sales:", round(df["Net Sales"].sum(), 2))

order_sales = df.groupby("Order ID")["Net Sales"].sum()
print("Average Order Value:", round(order_sales.mean(), 2))
print("Highest Order Sales:", round(order_sales.max(), 2))

print("\nSALES BY REGION")
print(df.groupby("Region")["Net Sales"].sum().sort_values(ascending=False))

print("\nSALES BY CHANNEL")
print(df.groupby("Sales Channel")["Net Sales"].sum().sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Replace any true nulls or string 'nan' representation in Region with 'Unknown'
df["Region"] = df["Region"].fillna("Unknown").astype(str).str.strip()
df["Region"] = df["Region"].replace({"nan": "Unknown", "Nan": "Unknown"})
df["Region"] = df["Region"].str.title()

# Re-calculate regional sales grouping
region_sales = df.groupby("Region")["Net Sales"].sum().sort_values(ascending=False)

# Plot the corrected regional breakdown
region_sales.plot(kind="bar", title="Sales by Region (Corrected)",
                  xlabel="Region", ylabel="Net Sales")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Strip leading and trailing whitespace from the 'Product' column to merge duplicates (e.g. 'Headphones ' and 'Headphones')
df["Product"] = df["Product"].astype(str).str.strip()

product_sales = df.groupby("Product")["Net Sales"].sum().sort_values(ascending=False)

product_sales.plot(kind="bar", title="Sales by Product (Corrected)",
                   xlabel="Product", ylabel="Net Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

monthly_sales = (
    df.dropna(subset=["Order Date"])
    .assign(Month=lambda x: x["Order Date"].dt.to_period("M"))
    .groupby("Month")["Net Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind="line", marker="o",
                   title="Monthly Sales Trend",
                   xlabel="Month", ylabel="Net Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
